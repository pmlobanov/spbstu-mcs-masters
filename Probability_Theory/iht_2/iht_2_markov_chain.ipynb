### Задание 6. Смоделировать траектории цепи Маркова на 10, 50, 100 и 1000 шагов, начинающиеся в различных состояниях для каждого случая

```python
class MarkovChain:
    def __init__(self, transition_matrix):
        """
            Метод инициализации класса MarkovChain.
            Параметры: transition_matrix (numpy.ndarray): Матрица переходов.
        """
        self.transition_matrix = transition_matrix

    def step(self, state):
        """
            Метод для выполнения одного шага в цепи Маркова.
            Параметры: state (int): Текущее состояние.
            Возвращает: int: Следующее состояние.
        """
        return int(np.random.choice(len(self.transition_matrix), p=self.transition_matrix[state]))


    def simulate(self, start_state, num_steps):
        """ 
            Метод для симуляции цепи Маркова.
            Параметры:
                start_state (int): Начальное состояние.
                num_steps (int): Количество шагов для симуляции.
            Возвращает: list: Список состояний, через которые прошла цепь.
        """
        states = [start_state]
        current_state = start_state
        for _ in range(num_steps):
            current_state = self.step(current_state)
            states.append(current_state)
        return states
```

```python
def count_probabilities(simulated_states:dict) -> dict:
    """
        Функция для подсчета вероятностей переходов между состояниями.
        Параметры: simulated_states (dict): Словарь с симулированными состояниями.
        Возвращает: dict: Словарь с вероятностями переходов.
    """
    probabilities = {}
    for key, states in simulated_states.items():
        start_state, num_steps = key
        counts = np.bincount(states, minlength=len(P))
        probabilities[key] = counts / num_steps
    return probabilities
```
